```python
import pandas as pd

# Group by the sub-dataset name and look at the 'properties.label' column
label_audit = dataset.labels.groupby('properties.dataset')['properties.label'].unique()

for sub_ds, labels in label_audit.items():
    # Filter out None/NaN to see if actual crop names exist
    clean_labels = [l for l in labels if l is not None and not (isinstance(l, float))]
    
    if len(clean_labels) > 0:
        print(f"✅ {sub_ds}: MULTICLASS (Types: {', '.join(clean_labels[:3])}...)")
    else:
        print(f"❌ {sub_ds}: BINARY ONLY")
```

```python
import pandas as pd

# 1. Group the labels to check for crop-specific names
# We check every unique entry in 'properties.label' for each sub-dataset
audit = dataset.labels.groupby('properties.dataset')['properties.label'].unique()

binary_subsets = []

for name, labels in audit.items():
    # A dataset is "Binary Only" if it contains NO specific crop strings
    has_multiclass = any(isinstance(l, str) and l.lower() != 'none' for l in labels)
    
    if not has_multiclass:
        # Get the count of samples for this binary dataset
        count = len(dataset.labels[dataset.labels['properties.dataset'] == name])
        binary_subsets.append({'Sub-Dataset': name, 'Samples': count})

# 2. Display the results in a clean table
binary_df = pd.DataFrame(binary_subsets).sort_values(by='Samples', ascending=False)

print("--- Binary-Only Sub-Datasets in CropHarvest ---")
print(binary_df.to_string(index=False))

print(f"\nTotal Binary Samples: {binary_df['Samples'].sum()}")
```

```python
from torchgeo.datasets import CropHarvest
from torch.utils.data import Subset

# 1. Load the full benchmark dataset
full_dataset = dataset

# 2. Find the indices where the sub-dataset is 'geowiki-landcover-2017'
# Note: Ensure the string matches exactly what appeared in your audit
target_name = 'geowiki-landcover-2017'
geowiki_indices = full_dataset.labels.index[
    full_dataset.labels['properties.dataset'] == target_name
].tolist()

# 3. Create the isolated subset
geowiki_dataset = Subset(full_dataset, geowiki_indices)

print(f"Full Dataset Size: {len(full_dataset)}")
print(f"Isolated GeoWiki Size: {len(geowiki_dataset)}")

# Verify the first sample
sample = geowiki_dataset[0]
print(f"Sample source: {sample}") # Should print 'geowiki-landcover-2017'
```

```python
# Check class balance in the subset
geowiki_labels = full_dataset.labels.iloc[geowiki_indices]
balance = geowiki_labels['properties.is_crop'].value_counts()
print("\nClass Balance in GeoWiki Subset:")
print(balance)
```

```python
import matplotlib.pyplot as plt

# 1. Extract the Lat/Lon for only the GeoWiki subset
geowiki_metadata = full_dataset.labels.iloc[geowiki_indices]

plt.figure(figsize=(5, 3))

# 2. Scatter plot of the coordinates
# s=1 makes points small; alpha=0.3 makes them semi-transparent
plt.scatter(
    geowiki_metadata['properties.lon'], 
    geowiki_metadata['properties.lat'], 
    c='forestgreen', 
    s=1, 
    alpha=0.1
)

plt.title(f"Geographic Distribution of '{target_name}' ({len(geowiki_indices)} points)")
plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.grid(True, linestyle='--', alpha=0.5)

# # Adding a rough outline of the world (optional but helpful)
# plt.axhline(0, color='black', linewidth=0.5) # Equator
# plt.axvline(0, color='black', linewidth=0.5) # Prime Meridian

plt.show()
```
